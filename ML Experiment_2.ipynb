```python
import pandas as pd
mushrooms_df = pd.read_csv('/content/mushrooms.csv')
display(mushrooms_df.describe())
```

```python
# Check unique values for all columns
for column in mushrooms_df.columns:
    print(f"Column '{column}': {mushrooms_df[column].nunique()} unique values")
    print(mushrooms_df[column].value_counts())
    print("\n" + "-"*30 + "\n")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Visualize the distribution of the target variable 'class'
fig = plt.figure(figsize=(6, 4))
sns.countplot(x='class', data=mushrooms_df, palette='viridis')
plt.title('Distribution of Mushroom Class (Edible vs. Poisonous)')
plt.xlabel('Class')
plt.ylabel('Count')
plt.show()
```

```python
fig = plt.figure(figsize=(10, 6))
sns.countplot(x='odor', hue='class', data=mushrooms_df, palette='viridis')
plt.title('Distribution of Odor by Mushroom Class')
plt.xlabel('Odor')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.legend(title='Class')
plt.tight_layout()
plt.show()
```

```python
fig = plt.figure(figsize=(12, 6))
sns.countplot(x='gill-color', hue='class', data=mushrooms_df, palette='viridis')
plt.title('Distribution of Gill Color by Mushroom Class')
plt.xlabel('Gill Color')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.legend(title='Class')
plt.tight_layout()
plt.show()
```
